# Load Data
Here I have taken 4 small regions from different sections of Xenium data from the marmoset brain

````python

ad1 = ad.read_h5ad("../data/CJ_BG_mini1.h5ad")
ad2 = ad.read_h5ad("../data/CJ_BG_mini2.h5ad")
ad3 = ad.read_h5ad("../data/CJ_BG_mini3.h5ad")
ad4 = ad.read_h5ad("../data/CJ_BG_mini4.h5ad")
````


### Start by creating the `SpatialCompare` object
### the anndata objects used in `spatial_compare` need a few things:

1. a category (column in .obs shared in both AnnData objects )
2. an `.obsm` key where spatial location data is stored
3. names for the datasets

This has some validation of shared genes and checks for the  shared `category` column in the `.obs`



````python
sc = SpatialCompare(ad1, ad2, data_names=["mini1", "mini2"], category='AIT15_Subclass_name')
````

### next we'll run spatial_compare
This generates (potentially many) plots:
1. a spatial plot of the 2 datasets, colored by the user-supplied category
2. population comparisons 
3. Gene expression plots for each group
4. Summary plots of transcript detection differences across cell types and genes

````python
d1d2_cells = [g for g in ad1.obs.AIT15_Subclass_name.unique() if "D1" in g or "D2" in g]
non_d1d2_cells = [g for g in ad1.obs.AIT15_Subclass_name.unique() if g not in d1d2_cells]
````

````python
# here we introduce some expression differences to test outlier detection
sc.ad_0[:,["ACHE"]] = 0.25*sc.ad_0[:,["ACHE"]].X
sc.ad_0[:,["PTMA"]] = 2.0*sc.ad_0[:,["PTMA"]].X
````

````python
# by using a high number of top genes, we can detect outliers in gene expression

sc.run_and_plot(category_values = d1d2_cells, min_mean_expression=.02, n_top_genes=500)

````

````python
sc.run_and_plot(category_values = non_d1d2_cells, min_mean_expression=.02, n_top_genes=500)
````

# Spatial Detection Scores
We can also look at how transcript detection varies across the data, guided by the statistics of expression for each group in a reference dataset. Here I create a reference from the 4 AnnData objects we loaded above, and 1 of the datasets is queried against that combined reference. Once all groups have been compared, the results (total detection z-score, difference in total counts and log10(detection ratio) are binned across the dataset to produce a spatial map of detection.  



````python
combined_reference = pd.concat([ad1.obs, ad2.obs, ad3.obs, ad4.obs], axis=0)
reference_detection_stats = (
    combined_reference
    .groupby("supercluster_name", observed=True)["transcript_counts"]
    .agg(["mean", "std"])
)
precalculated_means = reference_detection_stats["mean"]
precalculated_stds = reference_detection_stats["std"]

s = spatial_detection_scores(
    ad1.obs,
    precalculated_means.copy(),
    precalculated_stds.copy(),
    query_name="marmoset 1",
    n_bins=30,
    non_spatial=False,
)
````

The function `spatial_detection_scores` can also be called with `non_spatial=True` option that ignores the per-group statistics and just returns spatial maps compared to all cells in the reference. As seen below, this non-spatial version of relative detection is confused by the per-group variation in total transcripts, incorrectly highlighting a white matter region with fewer counts as a large negative z-score for detection.

````python
import numpy as np
from scipy import sparse

from spatial_compare.utils import create_test_data_from_spatial

# Use a bundled dataset that is available in CI. Cell 3 loads ad1 from ../data.
coordinate_columns = ["x_centroid", "y_centroid"]
category_column = "AIT15_Subclass_name"
coordinates = ad1.obs[coordinate_columns].to_numpy()
normalized_coordinates = (
    (coordinates - coordinates.min(axis=0))
    / np.ptp(coordinates, axis=0)
)
original_x = ad1.X.copy()

# All genes in the center square should be scaled exactly when noise is disabled.
size = 0.25
center_mask = np.all(np.abs(normalized_coordinates - 0.5) <= size / 2, axis=1)
assert center_mask.any() and (~center_mask).any()

center_result = create_test_data_from_spatial(
    ad1,
    category_column,
    spatial_coords=coordinate_columns,
    shape="square",
    size=size,
    location="center",
    loss_factor=0.5,
    noise_factor=0,
)
assert sparse.issparse(center_result.X)
assert np.allclose(
    center_result.X[center_mask].toarray(),
    ad1.X[center_mask].toarray() * 0.5,
)
assert (center_result.X[~center_mask] != ad1.X[~center_mask]).nnz == 0
assert (ad1.X != original_x).nnz == 0

# Test a corner circle with category and gene restrictions plus seeded noise.
circle_center = np.array([size / 2, 1 - size / 2])
circle_mask = (
    np.sum((normalized_coordinates - circle_center) ** 2, axis=1)
    <= (size / 2) ** 2
)
loss_category = ad1.obs.loc[circle_mask, category_column].value_counts().index[0]
category_mask = ad1.obs[category_column].eq(loss_category).to_numpy()
affected_mask = circle_mask & category_mask
assert affected_mask.any()

affected_x = ad1.X[affected_mask]
loss_gene_index = int(np.asarray(affected_x.sum(axis=0)).argmax())
loss_gene = ad1.var_names[loss_gene_index]
original_gene_counts = ad1.X[affected_mask, loss_gene_index].toarray().ravel()
assert np.any(original_gene_counts > 0)

np.random.seed(42)
expected_gene_counts = np.clip(
    np.random.normal(0.5 * original_gene_counts, 0.1 * original_gene_counts),
    0,
    None,
)
np.random.seed(42)
circle_result = create_test_data_from_spatial(
    ad1,
    category_column,
    spatial_coords=coordinate_columns,
    shape="circle",
    size=size,
    location="upper-left",
    loss_factor=0.5,
    noise_factor=0.1,
    loss_category_values=[loss_category],
    loss_genes=[loss_gene],
)
assert np.allclose(
    circle_result.X[affected_mask, loss_gene_index].toarray().ravel(),
    expected_gene_counts,
)
assert (
    circle_result.X[~affected_mask, loss_gene_index]
    != ad1.X[~affected_mask, loss_gene_index]
).nnz == 0
other_gene_indices = np.arange(ad1.n_vars) != loss_gene_index
assert (circle_result.X[:, other_gene_indices] != ad1.X[:, other_gene_indices]).nnz == 0
assert (ad1.X != original_x).nnz == 0

print(
    "All create_test_data_from_spatial checks passed using CJ_BG_mini1.h5ad "
    f"({center_mask.sum()} center cells; {affected_mask.sum()} "
    f"{loss_category} cells and gene {loss_gene} in the corner test)."
)
````

````python
s = spatial_detection_scores(
    circle_result.obs,
    precalculated_means.copy(),
    precalculated_stds.copy(),
    query_name="marmoset 1 with simulated loss",
    n_bins=30,
    non_spatial=False,
)
````

````python
# Create a strong, spatially localized loss across all genes.
np.random.seed(42)
strong_loss_data = create_test_data_from_spatial(
    ad1,
    "supercluster_name",
    spatial_coords=["x_centroid", "y_centroid"],
    shape="square",
    size=0.4,
    location="lower-right",
    loss_factor=0.5,
    noise_factor=0.1,
)

# spatial_detection_scores uses this obs column, so update it from the perturbed X.
strong_loss_data.obs["transcript_counts"] = np.asarray(
    strong_loss_data.X.sum(axis=1)
).ravel()

strong_loss_scores = spatial_detection_scores(
    strong_loss_data.obs,
    precalculated_means.copy(),
    precalculated_stds.copy(),
    query_name="marmoset 1 with strong  loss",
    n_bins=30,
    non_spatial=False,
)
````
